## 03. 2D Euclidean Rotation Matrices & Triangle Isometry
The 2D rotation matrix is given by:
$$R(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}$$
Demonstrating $R^T R = I_2$ and triangle angle sum $\Sigma = 180^\circ$.


```python
theta_deg = 45.0
R = euclidean_rotation_matrix(theta_deg)
R_T_R, max_diff = check_matrix_preservation(R, np.eye(2))

A_e = np.array([0.0, 0.0])
B_e = np.array([1.5, 0.0])
C_e = np.array([0.5, 1.2])

ang_A, ang_B, ang_C, sum_euc = euclidean_triangle_angles(A_e, B_e, C_e)

print("Rotation Matrix R(45 deg):")
print(R)
print("R^T R Deviation from I_2:", max_diff)
print(f"Euclidean Angles: A={ang_A:.1f} deg, B={ang_B:.1f} deg, C={ang_C:.1f} deg")
print(f"Euclidean Angle Sum: {sum_euc:.1f} deg (Always 180 deg)")
```

## 04. Spherical Representation ($p \in \mathbb{R}^3, p^T p = 1$) & Spherical Excess
Points on the unit sphere $S^2$ satisfy the constraint $x^2 + y^2 + z^2 = 1$.
Demonstrating that spherical triangle angle sum $\Sigma > 180^\circ$.


```python
A_s = latlon_to_cartesian(45.0, 0.0)
B_s = latlon_to_cartesian(0.0, 60.0)
C_s = latlon_to_cartesian(60.0, 30.0)

ang_As, ang_Bs, ang_Cs, sum_sph, excess_sph = spherical_triangle_angles(A_s, B_s, C_s)

print("Constraint A^T A =", dot_product(A_s, A_s))
print(f"Spherical Angles: A={ang_As:.1f} deg, B={ang_Bs:.1f} deg, C={ang_Cs:.1f} deg")
print(f"Spherical Angle Sum : {sum_sph:.1f} deg (> 180 deg)")
print(f"Spherical Excess E  : +{excess_sph:.1f} deg")
```

## 05. Hyperboloid Model ($p^T H p = -1$) & Angular Defect
In Minkowski 3-space $\mathbb{R}^{2,1}$, the hyperbolic plane is the upper sheet ($z > 0$) of the hyperboloid:
$$x^2 + y^2 - z^2 = -1 \quad \implies \quad p^T H p = -1, \quad H = \text{diag}(1, 1, -1)$$
Demonstrating that hyperbolic triangle angle sum $\Sigma < 180^\circ$.


```python
A_h = hyperboloid_point_from_disk(-0.4, 0.3)
B_h = hyperboloid_point_from_disk(0.5, -0.2)
C_h = hyperboloid_point_from_disk(0.0, -0.6)

ang_Ah, ang_Bh, ang_Ch, sum_hyp, defect_hyp = hyperbolic_triangle_angles(A_h, B_h, C_h)

print("Minkowski Form A^T H A:", minkowski_product(A_h, A_h))
print(f"Hyperbolic Angles: A={ang_Ah:.1f} deg, B={ang_Bh:.1f} deg, C={ang_Ch:.1f} deg")
print(f"Hyperbolic Angle Sum : {sum_hyp:.1f} deg (< 180 deg)")
print(f"Angular Defect D     : -{defect_hyp:.1f} deg")
```

## 06. Lorentz Boost Transformation ($B^T H B = H$)
Lorentz transformations preserve the Minkowski bilinear form: $B^T H B = H$.


```python
t_rapidity = 0.8
B_lorentz = lorentz_boost_x(t_rapidity)
A_h_trans = matrix_vector_mult(B_lorentz, A_h)

B_T_H_B, diff = check_matrix_preservation(B_lorentz, H_METRIC)

print("Lorentz Boost Matrix B(0.8):")
print(B_lorentz)
print("Preservation check B^T H B = H Max Dev:", diff)
print("Transformed Point A' Minkowski Form:", minkowski_product(A_h_trans, A_h_trans))
```
